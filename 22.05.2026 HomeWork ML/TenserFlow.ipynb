```python
import tensorflow as tf
```

```python
X = tf.constant([[0.05, 0.1]], dtype=tf.float32)
Y_true = tf.constant([[0.5]], dtype=tf.float32)
```

```python
input_size = 2
hidden_size = 2
output_size = 1
```

```python
W1 = tf.Variable(tf.random.normal([input_size, hidden_size]))
b1 = tf.Variable(tf.zeros([1, hidden_size]))

W2 = tf.Variable(tf.random.normal([hidden_size, output_size]))
b2 = tf.Variable(tf.zeros([1, output_size]))
```

```python
lambda_lr = 0.1
```

```python
for _ in range(1, 101):
  with tf.GradientTape() as tape:
    Z_hidden = tf.matmul(X, W1) + b1
    H = tf.sigmoid(Z_hidden)

    y_pred = tf.matmul(H, W2) + b2

    loss = tf.reduce_mean(tf.square(Y_true - y_pred))

  gradients = tape.gradient(loss, [W1, b1, W2, b2])
  grad_w1, grad_b1, grad_w2, grad_b2 = gradients

  W1.assign_sub(lambda_lr * grad_w1)
  b1.assign_sub(lambda_lr * grad_b1)
  W2.assign_sub(lambda_lr * grad_w2)
  b2.assign_sub(lambda_lr * grad_b2)
```

```python
mse = tf.reduce_mean(tf.square(Y_true - y_pred))
```

```python
print(f"Result: {y_pred.numpy()[0][0]} \nMSE: {mse}")
```
